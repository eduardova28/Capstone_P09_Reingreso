## Resumen ejecutivo

**Decisión que se busca mejorar.** Al momento del alta, el equipo clínico debe decidir qué pacientes diabéticos recibirán seguimiento reforzado. Actualmente la asignación depende de protocolos generales, criterio clínico y disponibilidad; no existe en las fuentes revisadas evidencia de una priorización individual sistemática basada en riesgo de reingreso.

**Problema central.** Con capacidad limitada, el equipo de alta no distingue de manera consistente qué episodios tienen mayor riesgo de reingreso antes de 30 días; por ello puede dejar sin intervención a pacientes de alto riesgo o consumir capacidad en pacientes de menor riesgo.

**Hipótesis de solución.** Un puntaje de riesgo calculado con información disponible antes o al momento del alta puede ordenar los episodios elegibles y apoyar —sin sustituir— el juicio clínico. La intervención prevista es una llamada temprana, revisión de adherencia/dieta y, cuando corresponda, una cita o plan de seguimiento.

**Capacidad de acción.** La consulta reporta aproximadamente **20 pacientes por turno de seis horas**.

**Condición de viabilidad.** El proyecto solo genera valor si: (a) el puntaje mejora una regla operativa simple; (b) el umbral se ajusta a la capacidad real; (c) no usa información posterior al alta; (d) se valida agrupando por paciente; y (e) el hospital puede ejecutar el seguimiento indicado.

**Estado al cierre de E1.** El problema y la acción están suficientemente definidos para continuar. Permanecen pendientes el costo institucional del reingreso y del seguimiento, la política operativa actual, el número semanal de cupos, la autoridad que aprobaría un piloto y algunos criterios clínicos de inclusión/exclusión.

## Salida 1 de 12 — Antecedentes

### Situación actual

- La base de investigación contiene 101 766 hospitalizaciones de 71 518 pacientes diabéticos atendidos en 130 hospitales de Estados Unidos entre 1999 y 2008; 11,2 % de los episodios registra reingreso antes de 30 días.
- La decisión ocurre al preparar el alta: asignar o no seguimiento reforzado.
- Un reingreso temprano es costoso y puede ser penalizado; parte de esos eventos podría evitarse mediante transición de cuidados, llamada o cita temprana.
- La Dra. Lares indicó que el Servicio/Comité de Salud Renal realiza seguimiento mediante llamadas, preguntas sobre cumplimiento diario, dieta y controles.
- Se reportó una capacidad aproximada de 20 pacientes por turno de seis horas.
- Cuatro o más hospitalizaciones previas en el año se consideran una señal clínicamente útil de alta probabilidad de reingreso.

### Árbol del problema

| Categoría | Descripción |
| :--- | :--- |
| **Causas raíz** | Capacidad de seguimiento limitada<br><br>información clínica fragmentada<br><br>ausencia de variables de adherencia y disponibilidad de insulina<br><br>datos históricos y de otro sistema sanitario |
| **Causas directas** | Priorización no individualizada<br><br>señales dispersas<br><br>criterios heterogéneos<br><br>dificultad para comparar riesgo entre pacientes |
| **Problema central** | El equipo de alta no distingue consistentemente qué pacientes diabéticos elegibles tienen mayor riesgo de reingreso antes de 30 días |
| **Efectos** | Pacientes de alto riesgo sin seguimiento<br><br>cupos usados en menor riesgo<br><br>reingresos potencialmente evitables<br><br>mayor costo y carga asistencial |

### Proceso actual y brecha

La especialista refiere un protocolo guiado por metas clínicas, pero las fuentes no documentan un instrumento de priorización individual ni un umbral operativo común. El modelo propuesto no reemplaza guías ni juicio médico: añade una estimación comparable para ordenar una cola de pacientes cuando los cupos son escasos.

### Flujo de decisión futuro

`historia disponible antes del alta → elegibilidad → riesgo estimado → lista priorizada → confirmación clínica → seguimiento → resultado a 30 días`

## Salida 2 de 12 — Objetivos del negocio

### Objetivo principal

**Mejorar la asignación de la capacidad limitada de seguimiento posalta, priorizando a pacientes diabéticos elegibles con mayor riesgo de reingreso antes de 30 días, para reducir reingresos potencialmente evitables sin exceder los cupos operativos ni desplazar el juicio clínico.**

### Objetivos específicos

1. Entregar al equipo de alta una lista diaria o semanal ordenada por riesgo y limitada a su capacidad real.
2. Aumentar la proporción de pacientes que reingresarían antes de 30 días capturados dentro de los cupos de seguimiento frente a la política actual y frente a una regla simple basada en hospitalizaciones previas.
3. Hacer explícitos los factores que sustentan cada prioridad para permitir confirmación o ajuste clínico.
4. Evitar recomendaciones sobre episodios no accionables, por ejemplo fallecimientos durante el episodio índice.
5. Monitorear el desempeño por grupos relevantes y evitar utilizar raza como factor de priorización clínica directa.

### Usuario e intervención

- **Usuario principal:** enfermera de enlace o personal del servicio encargado del seguimiento.
- **Acción:** llamada temprana, comprobación de cumplimiento y dieta, coordinación de cita/control y escalamiento según protocolo.
- **Beneficiario:** paciente diabético dado de alta y su red de cuidado.
- **Producto conceptual:** checklist/lista de alta inteligente con riesgo, factores explicativos y nivel de seguimiento sugerido.

## Salida 3 de 12 — Criterios de éxito del negocio

### Criterio SMART propuesto para el piloto

> **Durante un piloto prospectivo de 12 semanas, y usando como máximo la capacidad semanal confirmada del equipo, incrementar en al menos 15 % relativo la captación de pacientes que reingresan antes de 30 días dentro del grupo asignado a seguimiento, frente a la política operativa vigente, sin aumentar el número de contactos programados; adicionalmente, reducir en al menos 10 % relativo el reingreso a 30 días en los pacientes seguidos frente a un comparador acordado.**

Los porcentajes y la duración son **umbrales de decisión iniciales**, no resultados observados. Deben aprobarlos el dueño del proceso y el pagador antes de un piloto.

| Elemento SMART | Definición |
|---|---|
| Específico | Priorizar seguimiento de pacientes diabéticos elegibles al alta |
| Medible | Captación dentro de cupos, reingreso a 30 días, contactos realizados y uso de capacidad |
| Alcanzable | La intervención ya existe conceptualmente; se reorganiza a quién se dirige |
| Relevante | Reduce eventos evitables, carga clínica y costo potencial |

### Baselines de negocio

1. **Principal:** política vigente del hospital (criterio/protocolo actual y su captación histórica).
2. **Comparador reproducible:** priorizar por número de hospitalizaciones previas en el año, señal respaldada por la consulta.
3. **Referencia poblacional:** prevalencia global de reingreso <30 días cercana a 11,2 %; no sustituye al baseline operativo.

### Guardas de éxito

- No exceder cupos ni horas disponibles.
- No reducir la cobertura de pacientes definidos como obligatorios por guía.
- Auditar diferencias de captación y error por sexo, edad y otros grupos clínicamente pertinentes.
- El dueño del proceso y el pagador juzgarán el criterio de negocio; ambos cargos concretos están * pendientes** de confirmación.

# 1.3 Determinar los objetivos de minería de datos

## Salida 9 de 12 — Objetivos de minería de datos

### Objetivo técnico

Desarrollar un modelo de **clasificación binaria probabilística** que estime, para cada episodio elegible, la probabilidad de reingreso registrado antes de 30 días y permita ordenar pacientes de acuerdo con una capacidad K configurable.

### Definición del problema

| Elemento | Definición previa al modelado |
|---|---|
| Sujeto | Episodio de hospitalización elegible; un paciente puede aportar varios episodios |
| Momento de predicción | Antes o durante el cierre del alta, previo a asignar seguimiento |
| Etiqueta positiva | `readmitted = <30` |
| Etiqueta negativa | `readmitted = >30` o `NO`, tras aplicar elegibilidad |
| Observación | Datos demográficos, administrativos y clínicos conocidos hasta el alta |
| Resultado | Reingreso observado en los 30 días posteriores |
| Salida | Probabilidad calibrada, nivel de riesgo, factores explicativos y prioridad |
| Uso | Ordenar cupos; no diagnosticar ni reemplazar al profesional |

### Población y elegibilidad inicial

- **Excluir:** episodios cuyo destino indica fallecimiento (ID 11 y 19–21), porque no pueden experimentar el resultado.
- **Revisar:** hospicio, transferencias, rehabilitación, psiquiatría y destinos desconocidos, por accionabilidad/captura diferencial.
- **Separar o excluir:** altas voluntarias, debido a mecanismo y riesgo distintos.
- **Conservar para auditoría:** raza, pero no como predictor directo en la primera especificación.

### Variables y conocimiento clínico

- Comparar códigos diagnósticos crudos con agrupación clínica versionada.
- Candidatos a grupo propio: insuficiencia cardíaca, piel/celulitis, neurológicas, infecciosas y subgrupos de diabetes.
- Hospitalizaciones previas son señal candidata y baseline operativo.
- Ausencia de HbA1c, glucosa o peso puede representar proceso de atención; no se interpretará causalmente ni se imputará por defecto.
- Disponibilidad y aceptación de insulina constituyen confusores/variables omitidas relevantes.

### Prevención de fuga

Cada variable responderá: **¿estaba disponible de forma confiable antes de asignar seguimiento?** Si no, se excluye. El código de destino puede definir elegibilidad, pero no debe convertirse automáticamente en predictor sin evaluación temporal y clínica.

## Salida 10 de 12 — Criterios de éxito de minería de datos

### Esquema de validación declarado antes de entrenar

1. Separación por `patient_nbr` mediante validación cruzada agrupada o holdout agrupado.
2. Si la información lo permite, validación adicional por hospital/periodo; de no existir identificador de sitio o fecha exacta, declarar la limitación.
3. Todo preprocesamiento se ajusta solo con entrenamiento.
4. Reporte con intervalos de confianza y por subgrupos.

### Baselines técnicos

- **B0:** probabilidad constante igual a prevalencia de entrenamiento.
- **B1:** regresión logística regularizada con variables disponibles al alta.
- **B2:** regla simple por hospitalizaciones previas en el año, acordada con clínica.

### Métrica primaria y umbrales

**Métrica primaria:** `Recall@K`, donde K es la capacidad confirmada del equipo en el mismo horizonte de priorización. Esta métrica mide qué proporción de reingresos se captura en los cupos realmente accionables.

**Criterio técnico inicial:**

- superar en al menos **10 % relativo** el `Recall@K` del mejor baseline;
- mantener **AUCPR ≥ 1,5 veces** la prevalencia del conjunto de prueba;
- presentar calibración con pendiente/intercepto e índice Brier, sin deterioro sustantivo frente al baseline;
- no mostrar una diferencia absoluta de sensibilidad mayor a **10 puntos porcentuales** entre grupos con tamaño suficiente, o documentar y mitigar la causa.

### Métricas secundarias

`Precision@K`, AUCPR, ROC-AUC, sensibilidad, especificidad, Brier score, curva de calibración y matriz de confusión al umbral operativo.

### Política de umbral

No se usará 0,5 por defecto. Se seleccionará el top K o un umbral que respete la capacidad. Si clínica confirma que el falso negativo es claramente más costoso, se añadirá una restricción mínima de sensibilidad.

### Técnicas que no se adoptan aún

- No se elige un algoritmo final antes del EDA y los baselines.
- No se imputan faltantes extremos por regla general.
- No se usa sobremuestreo antes de definir particiones y calibración.
- No se interpreta asociación como causalidad ni el score como beneficio esperado de intervención.
- No se automatiza la asignación sin revisión humana y monitoreo.

### Criterio de selección final

Se preferirá el modelo que maximice valor operativo a K, esté bien calibrado, sea estable entre particiones/pacientes, no presente brechas injustificadas y pueda explicarse al usuario. Una mejora marginal de AUROC no compensará mala calibración, fuga o baja adopción.
